  问题1

   * 原文: 1) Describe how Gibbs free energy and critical radius related to each other. Derive the mathematical description. Comment on the passages and final results 
     (!)
   * 中文翻译: 1) 描述吉布斯自由能（Gibbs free energy）和临界半径（critical radius）是如何相互关联的。推导出它们之间的数学关系式。请对推导过程和最终结果进行评述。

  核心概念理解:

  想象一下在饱和的水蒸气中，要形成一个微小的液滴（这个过程叫成核，nucleation）。这个过程会发生两种能量变化：

   1. 不利变化（能量增加）： 凭空创造出一个液滴，意味着创造了新的“表面”（液体和蒸气的交界面）。维持这个表面需要能量，就像吹泡泡需要力气一样。这个能量叫做表面能，它与液滴
      的表面积（4πr²）成正比。这部分能量变化是正值（ΔG_表面 > 0），阻碍液滴形成。
   2. 有利变化（能量降低）： 
      蒸气分子凝聚成液体，会释放能量，因为液体状态比过饱和蒸气状态更稳定。这个能量叫做体能量，它与液滴的体积（(4/3)πr³）成正比。这部分能量变化是负值（ΔG_体积 < 
      0），驱动液滴形成。

  总的吉布斯自由能变化 ΔG 就是这两部分的和。液滴的半径 r 决定了这两部分能量的相对大小。临界半径 `r*` 就是一个临界点：
   * 当液滴半径 r < r* 时，表面能的增加占主导，液滴不稳定，会倾向于蒸发消失。
   * 当液滴半径 r > r* 时，体能量的降低占主导，液滴会稳定存在并继续长大。

  数学推导:

   1. 总吉布斯自由能变化 (Total Gibbs Free Energy Change), ΔG(r):
      ΔG(r) = ΔG_表面 + ΔG_体积

   2. 表面能项 (Surface Term):
      ΔG_表面 = 4πr²γ
       * r 是液滴半径。
       * γ (gamma) 是表面张力（surface tension），单位面积的表面能。

   3. 体能量项 (Volume Term):
      ΔG_体积 = (4/3)πr³ΔG_v
       * ΔG_v 是单位体积的相变自由能变化。它是一个负值，代表相变的驱动力。ΔG_v 与过饱和度 S (Supersaturation) 相关：
          ΔG_v = - (kT/v) * ln(S)
           * k 是玻尔兹曼常数。
           * T 是绝对温度。
           * v 是液体中一个分子的体积。
           * S = P/P₀ 是过饱和度，即蒸气实际分压 P 与该温度下饱和蒸气压 P₀ 的比值。只有当 S > 1 时，ln(S) 才为正，ΔG_v 才为负，才有成核的驱动力。

   4. 组合起来:
      ΔG(r) = 4πr²γ - (4/3)πr³ * (kT/v) * ln(S)

   5. 求临界半径 (r*):
      临界半径 r* 对应于 ΔG(r) 函数的极大值点（能量壁垒的顶峰）。我们对 ΔG(r) 求关于 r 的导数，并令其等于零：
      d(ΔG)/dr = 8πrγ - 4πr² * (kT/v) * ln(S) = 0
      解这个方程，得到临界半径 r*：
      r* = (2γv) / (kT * ln(S))

   6. 求临界能垒 (ΔG*):
      将 r* 的表达式代回到 ΔG(r) 的方程中，得到形成一个稳定核所需的能量壁垒 ΔG*：
      ΔG* = (16πγ³v²) / (3 * (kT * ln(S))²)
      也可以简化为：
      ΔG* = (4/3)π(r*)²γ

  评述 (Comment):

   * 关系: 吉布斯自由能 ΔG 的变化曲线解释了为什么存在一个“临界”半径。成核过程不是自动发生的，它需要克服一个能量壁垒 ΔG*。
   * 临界半径 `r*` 的意义: 它代表了能够稳定存在的最小核的尺寸。这个尺寸由材料本身的性质（表面张力 γ，分子体积 v）和外界条件（温度 T，过饱和度 S）共同决定。
   * 临界能垒 `ΔG*` 的意义: 它是成核过程的“启动资金”。系统需要吸收这么多能量（通常来自随机的热涨落）才能形成一个稳定的核。ΔG* 越高，成核就越困难。
   * 过饱和度的影响: 从公式可以看出，过饱和度 S 越大，ln(S) 越大，则临界半径 r* 和临界能垒 ΔG* 都越小。这意味着，驱动力越强，成核越容易，所需要的初始核尺寸也越小。

  定性关系图解析:

  你需要画一个坐标系，X轴是半径 (r)，Y轴是吉布斯自由能变化 (ΔG)。

   1. 情况一: S < 1 (未饱和, Undersaturated)
       * 此时 ln(S) 为负值，导致体能量项 ΔG_体积 变为正值。
       * ΔG(r) = 4πr²γ + (正数) * r³
       * 表面能和体能量项都是正的。所以 ΔG 会随着 r 的增加而单调递增。
       * 图形: 一条从原点 (0,0) 出发，持续向上攀升的曲线。
       * 物理意义: 系统中没有任何驱动力去形成新的相，任何偶然形成的微小团簇都会立即消失。不会发生自发成核。

   2. 情况二: S > 1 (过饱和, Supersaturated)
       * 此时 ln(S) 为正值，ΔG_体积 为负值。
       * ΔG(r) = 4πr²γ - (正数) * r³
       * 当 r 很小时，r² 项占主导，曲线随 r 增加而上升。
       * 当 r 很大时，r³ 项占主导，曲线随 r 增加而下降。
       * 图形: 一条从原点 (0,0) 出发，先上升到一个峰值（顶点），然后开始下降的曲线。这个峰值点对应的横坐标就是临界半径 `r*`，纵坐标就是临界能垒 `ΔG*`。

   3. 情况三: S > 2 (更高程度的过饱和)
       * S > 2 只是 S > 1 的一个特例，但饱和度更高。
       * 根据我们推导的公式，S 越大，r* 和 ΔG* 都越小。
       * 图形: 与 S > 1 的曲线形状相同，但是峰值更低，且峰值位置更靠左。
       * 物理意义: 过饱和度越高，成核的能量壁垒越低，需要的临界尺寸也越小，成核越容易发生。

  问题3

   * 原文: 3) Draw a qualitative graph of the Gibbs free energy and critical radius when the surface tension of the vapors compound is halved.
   * 中文翻译: 3) 当蒸气化合物的表面张力减半时，画出吉布斯自由能随半径变化的定性关系图。

  解析:

  这个问题是基于 S > 1 的情况，比较表面张力 γ 和 γ/2 的区别。

  我们再次看我们的关键公式：
   * r* = (2γv) / (kT * ln(S))
   * ΔG* = (16πγ³v²) / (3 * (kT * ln(S))²)

  当 γ 变为 γ/2 时：

   1. 对临界半径 `r*` 的影响:
      r* 与 γ 成正比。所以，如果 γ 减半，新的临界半径 `r*` 也将减半。

   2. 对临界能垒 `ΔG*` 的影响:
      ΔG* 与 γ³ (gamma的立方) 成正比。所以，如果 γ 减半，新的临界能垒 `ΔG*` 将变为原来的 (1/2)³ = 1/8。能量壁垒会急剧降低！

  定性图 (Qualitative Graph):

  在同一张图上画出两条 S > 1 的曲线。

    1       ΔG (Gibbs Free Energy)
    2        ^
    3        |
    4        |
    5        |      / . . . . . . . . . . . . . . . (Original, γ)
    6        |     /                             .
    7        |    * (Original ΔG*)               .
    8        |   / \                             .
    9        |  /   \
   10        | * (New ΔG* = Original/8)
   11        |/ \   \ . . . . . . . . . . . . . . (New, γ/2)
   12       -+---|----*---------------------------> r (Radius)
   13        (0,0)  |    |
   14             New r*  Original r*
   15             =Original/2

  物理意义: 表面张力 γ 是形成新表面的阻力。当这个阻力减半时，成核过程会变得极其容易。不仅稳定核所需要的尺寸减半了，更重要的是，需要克服的能量壁垒降低到了原来的1/8。在
  实际应用中，加入表面活性剂（surfactant）就是一种降低表面张力的有效方法，可以促进液滴（如乳液）的形成和稳定。


```python
import numpy as np
import matplotlib.pyplot as plt

# --- Setup and Definitions ---

# Let's define arbitrary but consistent physical constants for plotting.
# We can group (kT/v) into a single constant for simplicity.
# Let's assume gamma_val = 1.0 and (kT/v) = 1.0 for the base case.
# So, ΔGv = -ln(S) and r* = 2*gamma / ln(S)

def calculate_delta_g(r, gamma, ln_S):
    """Calculates Gibbs Free Energy change ΔG(r)."""
    surface_energy = 4 * np.pi * r**2 * gamma
    # The volume term is positive if ln_S is negative (S < 1)
    volume_energy = -(4/3) * np.pi * r**3 * ln_S
    return surface_energy + volume_energy

def calculate_critical_values(gamma, ln_S):
    """Calculates critical radius (r*) and critical energy (ΔG*)."""
    # Critical values only exist for S > 1 (ln_S > 0)
    if ln_S <= 0:
        return None, None
    r_star = (2 * gamma) / ln_S
    # Using the simplified formula ΔG* = (4/3)π(r*)²γ
    g_star = (4/3) * np.pi * r_star**2 * gamma
    return r_star, g_star

# Generate a range of radii for the x-axis
r = np.linspace(0.01, 8, 500)

# --- Plot for Question 2: Effect of Supersaturation (S) ---

fig1, ax1 = plt.subplots(figsize=(10, 6))

# Define cases for S
cases_s = {
    "S < 1 (e.g., S=0.5)": 0.5,
    "S > 1 (e.g., S=2.0)": 2.0,
    "S > 2 (e.g., S=4.0)": 4.0,
}
gamma_base = 1.0

for label, s_val in cases_s.items():
    ln_S = np.log(s_val)
    delta_g_values = calculate_delta_g(r, gamma_base, ln_S)
    ax1.plot(r, delta_g_values, label=label)

    # Annotate critical points for S > 1
    r_star, g_star = calculate_critical_values(gamma_base, ln_S)
    if r_star:
        ax1.plot(r_star, g_star, 'ko') # Black dot
        ax1.annotate(f'(r*, ΔG*)', xy=(r_star, g_star), xytext=(r_star + 0.5, g_star + 5),
                     arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=5))

# Formatting the plot
ax1.set_title('Question 2: Effect of Supersaturation (S) on Gibbs Free Energy', fontsize=14)
ax1.set_xlabel('Radius (r)', fontsize=12)
ax1.set_ylabel('Gibbs Free Energy Change (ΔG)', fontsize=12)
ax1.axhline(0, color='gray', linestyle='--', linewidth=1)
ax1.set_ylim(-10, 60)
ax1.set_xlim(0, 8)
ax1.legend()
ax1.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()

```

```python

# --- Plot for Question 3: Effect of Surface Tension (γ) ---

fig2, ax2 = plt.subplots(figsize=(10, 6))

# Define cases for gamma, keeping S constant
s_fixed = 2.0
ln_S_fixed = np.log(s_fixed)
cases_gamma = {
    "Original (γ = 1.0)": 1.0,
    "Halved (γ = 0.5)": 0.5,
}

for label, gamma_val in cases_gamma.items():
    delta_g_values = calculate_delta_g(r, gamma_val, ln_S_fixed)
    ax2.plot(r, delta_g_values, label=label)

    # Annotate critical points
    r_star, g_star = calculate_critical_values(gamma_val, ln_S_fixed)
    if r_star:
        ax2.plot(r_star, g_star, 'ko') # Black dot
        ax2.annotate(f'(r*, ΔG*)', xy=(r_star, g_star), xytext=(r_star + 0.5, g_star + 2),
                     arrowprops=dict(facecolor='black', shrink=0.05, width=1, headwidth=5))

# Formatting the plot
ax2.set_title('Question 3: Effect of Surface Tension (γ) on Gibbs Free Energy (S=2.0)', fontsize=14)
ax2.set_xlabel('Radius (r)', fontsize=12)
ax2.set_ylabel('Gibbs Free Energy Change (ΔG)', fontsize=12)
ax2.axhline(0, color='gray', linestyle='--', linewidth=1)
ax2.set_ylim(-5, 40)
ax2.set_xlim(0, 6)
ax2.legend()
ax2.grid(True, linestyle='--', alpha=0.6)

plt.tight_layout()
plt.show()
```
